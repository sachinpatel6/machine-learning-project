```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
data["DeviceType"]=le.fit_transform(data["DeviceType"])
```

```python
from sklearn.preprocessing import StandardScaler #scaling library

rs=StandardScaler()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x=data.drop("SmartHomeEfficiency",axis=1) # spltting to target and Features variable
y=data["SmartHomeEfficiency"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42 )
```

```python
# Apply Standard Scaler on numeric columns only

numeric_cols = ['UserID', 'UsageHoursPerDay', 'EnergyConsumption',
                'UserPreferences', 'MalfunctionIncidents', 'DeviceAgeMonths']

# Fit scaler on training data only
x_train[numeric_cols] = rs.fit_transform(x_train[numeric_cols])

# Transform test data
x_test[numeric_cols] = rs.transform(x_test[numeric_cols])
```

```python
#desicion tree
from sklearn.tree import DecisionTreeClassifier
dt=DecisionTreeClassifier()
dt.fit(x_train,y_train)
y_pred_dtc=dt.predict(x_test)

from sklearn.metrics import accuracy_score,confusion_matrix

print("accuracy",accuracy_score(y_test,y_pred_dtc))
```

```python
cm=confusion_matrix(y_test,y_pred_dtc)
```

```python
sns.heatmap(cm,annot=True,fmt="d")
plt.title("Confusion Matrix - Desicion tree")
plt.show()
```

```python
#random forest
from sklearn.ensemble import RandomForestClassifier
rf=RandomForestClassifier()
rf.fit(x_train,y_train)
y_pred_rf=rf.predict(x_test)

from sklearn.metrics import accuracy_score,confusion_matrix

print("accuracy",accuracy_score(y_test,y_pred_rf))
```

```python
cm=confusion_matrix(y_test,y_pred_rf)
```

```python
sns.heatmap(cm,annot=True,fmt="d")
plt.title("Confusion Matrix - Random Forest")
plt.show()
```

```python
#Xg boost
from xgboost import XGBClassifier
xg=XGBClassifier()
xg.fit(x_train,y_train)
y_pred_xg=xg.predict(x_test)

from sklearn.metrics import accuracy_score,confusion_matrix

print("accuracy",accuracy_score(y_test,y_pred_xg))
```

```python
cm=confusion_matrix(y_test,y_pred_xg)
```

```python
sns.heatmap(cm,annot=True,fmt="d")
plt.title("Confusion Matrix - Xg Boost")
plt.show()
```

```python
# Logistic Regression
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix

# increase iteration limit to avoid convergence warning
lr = LogisticRegression(max_iter=1000, solver='lbfgs')
lr.fit(x_train, y_train)
y_pred_lr = lr.predict(x_test)

# Evaluate
print("Accuracy:", accuracy_score(y_test, y_pred_lr))

cm = confusion_matrix(y_test, y_pred_lr)
sns.heatmap(cm, annot=True, fmt="d")
plt.title("Confusion Matrix - Logistic Regression")
plt.show()
```

```python
# Support Vector Machine (SVM) Classifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix

# Initialize SVM model
svm = SVC(kernel='rbf', C=1.0, gamma='scale', random_state=42)

# Train the model
svm.fit(x_train, y_train)

# Predict on test data
y_pred_svm = svm.predict(x_test)

# Evaluate performance
print("Accuracy:", accuracy_score(y_test, y_pred_svm))

# Confusion matrix visualization
cm = confusion_matrix(y_test, y_pred_svm)
sns.heatmap(cm, annot=True, fmt="d")
plt.title("Confusion Matrix - SVM Classifier")
plt.show()
```

```python
models = ['Decision Tree', 'Random Forest', 'XGBClassifier', 'Logistic Regression', 'SVM']

scores = [
    accuracy_score(y_test, y_pred_dtc),   # Decision Tree
    accuracy_score(y_test, y_pred_rf),    # Random Forest
    accuracy_score(y_test, y_pred_xg),    # XGBoost
    accuracy_score(y_test, y_pred_lr),    # Logistic Regression
    accuracy_score(y_test, y_pred_svm)    # SVM
]

for name, score in zip(models, scores):
    print(f'{name}: {score*100:.2f}%')
```
