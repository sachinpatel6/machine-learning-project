```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
```

```python
plt.figure(figsize=(20,10))
sns.heatmap(data.corr(),annot=True) # for showing no.s annot=True
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler #scaling library
rs=StandardScaler()
data["Is Daylight"]=le.fit_transform(data["Is Daylight"])
```

```python
from sklearn.model_selection import train_test_split
```

```python
x=data.drop("Power Generated",axis=1) # spltting to target and Features variable
y=data["Power Generated"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42 )
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()

 # fitting model
lr.fit(x_train,y_train)
```

```python
# predicting
y_pred=lr.predict(x_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
mae=mean_absolute_error(y_test,y_pred)
mse=mean_squared_error(y_test,y_pred)
r2=r2_score(y_test,y_pred)
```

```python
print("MAE is",mae)
print("MSE is",mse)
print("R2 is",r2)
```

```python
# decision tree
from sklearn.tree import DecisionTreeRegressor
dtr=DecisionTreeRegressor ()
dtr.fit(x_train,y_train)

y_pred_dtr=dtr.predict(x_test)
```

```python
mae=mean_absolute_error(y_test,y_pred_dtr)
mse=mean_squared_error(y_test,y_pred_dtr)
r2=r2_score(y_test,y_pred_dtr)
```

```python
print("MAE is",mae)
print("MSE is",mse)
print("R2 is",r2)
```

```python
# Random Tree
from sklearn.ensemble import RandomForestRegressor
rfr=RandomForestRegressor(n_estimators=100,random_state=42)
rfr.fit(x_train,y_train)

y_pred_rfr=rfr.predict(x_test)
```

```python
mae=mean_absolute_error(y_test,y_pred_rfr)
mse=mean_squared_error(y_test,y_pred_rfr)
r2=r2_score(y_test,y_pred_rfr)
```

```python
print("MAE is",mae)
print("MSE is",mse)
print("R2 is",r2)
```

```python
# for XG boost
from sklearn.ensemble import GradientBoostingRegressor
gbr=GradientBoostingRegressor()
gbr.fit(x_train,y_train)

y_pred_gbr=gbr.predict(x_test)
```

```python
mae=mean_absolute_error(y_test,y_pred_gbr)
mse=mean_squared_error(y_test,y_pred_gbr)
r2=r2_score(y_test,y_pred_gbr)
```

```python
print("MAE is",mae)
print("MSE is",mse)
print("R2 is",r2)
```
